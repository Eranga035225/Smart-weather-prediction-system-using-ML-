```python
import pandas as pd
```

```python
# Load dataset
df = pd.read_csv("../../data/weatherData.csv", encoding='utf-8')
```

```python
df.head()
```

```python
# Rename columns for easier use
df.rename(columns={
    'temperature_2m_mean (°C)': 'temperature',
    'precipitation_sum (mm)': 'precipitation',
    'rain_sum (mm)': 'rainfall',
    'wind_speed_10m_max (km/h)': 'wind_speed',
    'date': 'date' 
}, inplace=True)
```

```python
# Combine year, month, day into a datetime column
if 'date' not in df.columns:
    df['date'] = pd.to_datetime(df[['year', 'month', 'day']])
```

```python
# Drop unwanted columns
drop_cols = [
    "sunrise", "sunset", "shortwave_radiation_sum (MJ/m²)",
    "et0_fao_evapotranspiration (mm)",
    "wind_gusts_10m_max (km/h)", "wind_direction_10m_dominant (°)"
]
df.drop(columns=drop_cols, inplace=True, errors="ignore")
```

```python
# Drop duplicate records
df.drop_duplicates(inplace=True)
```

```python
# Drop rows with missing values in important target columns
essential_cols = ['temperature', 'rainfall', 'wind_speed', 'precipitation']
df.dropna(subset=essential_cols, inplace=True)
```

```python
#  Fill remaining missing values
df.ffill(inplace=True)
df.bfill(inplace=True)
```

```python
# Handle invalid/outlier values
# Replace negative rainfall/precipitation values with 0
df['rainfall'] = df['rainfall'].apply(lambda x: 0 if x < 0 else x)
df['precipitation'] = df['precipitation'].apply(lambda x: 0 if x < 0 else x)
```

```python
# Remove unrealistic temperatures
df = df[(df['temperature'] > -50) & (df['temperature'] < 60)]
```

```python
# Wind speed realistic range (0-150 km/h)
df = df[(df['wind_speed'] >= 0) & (df['wind_speed'] <= 150)]
```

```python
# Save cleaned dataset
df.to_csv("cleaned_data.csv", index=False)
print("Step 4: Data Cleaning complete.")
print(f"Shape after cleaning: {df.shape}")
print("Date range:", df['date'].min(), "to", df['date'].max())
```
